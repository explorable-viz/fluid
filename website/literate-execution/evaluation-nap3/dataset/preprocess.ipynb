```python
import pandas as pd

csv_path = "ccra3-actions-for-nap3.csv" # exported from Excel worksheet
json_path = "ccra3-actions-for-nap3.json"

df = pd.read_csv(
  csv_path,
  keep_default_na=False, # these two needed to map empty CVS cells to ""
  na_filter=False
)

# Excel worksheet range is longer than the data, so export generates blank rows; drop
df = df[~(df.eq("").all(axis=1))]

print(df.columns.tolist())
```

```python
df = df.rename(
    columns={
        "Clear actions appropriate for national adaptation policy programmes included in the assessment that would make a difference to risk score within the next five years\n": "Actions",
        "Action aimed at:\n• Government and/or ALB\n• Non-government\n• Unclear": "Action aimed at",
        "Typology\n• No and low regret adaptation, including capacity building\n• Climate-smart design, considering uncertainty\n• Early activities to support future adaptation\n• Further investigation\n• Unclear typology noted in text": "Typology",
        "Typology confidence \n• High\n• Medium\n• Low": "Typology confidence",
        "Evaluation of the extent to which NAP3 has addressed CCRA3 Actions, as set out in:\n• The NAP3 document [NAP3]\n• The risk-by-risk annex [RBRA]\n• The Government’s response to the CCC’s 2023 progress report [GRPR]": "Extent addressed",
        "Action linked with policies not listed within NAP3 documents": "Non-NAP3 policies",
        "Evaluation score:\n• ‘Fully addressed’\n• ‘Significantly addressed’\n• ‘Minimally addressed’\n• 'Recognised but not addressed’\n• ‘Not addressed’ ": "Evaluation score",
    }
)
print(df.columns.tolist())
```

```python
print("drop all columns derived via VLOOKUP join with ccra3-risks tab:")
# last non-derived column
n = df.columns.tolist().index('Action Owner')
cols = df.columns[n+1:].tolist()
print(cols)

df = df.drop(columns=cols)
print(df.columns.tolist())
```

```python
df.to_json(json_path, orient="records", indent=2, force_ascii=False)
print(f"Wrote {len(df)} records to {json_path}")
```

```python
csv_path = "ccra3-risks.csv" # exported from Excel worksheet
json_path = "ccra3-risks.json"

df = pd.read_csv(
  csv_path,
  keep_default_na=False, # these two needed to map empty CVS cells to ""
  na_filter=False
)

print(df.columns.tolist())
```

```python
df = df.rename(
    columns={
        "Risk or Opportunity Descriptor (BOLD = Committee priority risks)": "Risk or Opportunity Descriptor",
        "Committee Priority risk": "CCC Priority",
        "Key policy opportunities in the next 2 years (UK wide, Committee View)": "Key policy opportunities in next 2 years",
    }
)

df = df.drop(columns=[
    "Benefits of further adaptation action in the next five years (CCRA3 section X.X.3)",
    " Indicative costs and benefits of additional adaptation (CCRA3 section X.X.3.1)",
    "Unnamed: 33", # probably worksheet range extends to right of data as well, so export generates empty column
])

# worksheet range is longer than the data, so export generates 2 blank rows; drop
df = df[~(df.eq("").all(axis=1))]

print(df.columns.tolist())
```

```python
df.to_json(json_path, orient="records", indent=2, force_ascii=False)
print(f"Wrote {len(df)} records to {json_path}")
```
